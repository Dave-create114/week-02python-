```python
#Evaluates a numerical score and returns the corresponding letter grade.
try:
    # Prompt user for input and convert it to an integer
    score = int(input("Enter your score: "))
except ValueError:
    print("Please enter a valid integer score.")
    score = None

def check_grade(score):
    if score is None:
        return 'Invalid Score'
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    else:
        return 'F'

if score is not None:
    print(score)
    print(f"Your grade is: {check_grade(score)}")
```

```python
total = 0

print("Enter item prices. Type 'done' when you are finished.")

while True:
    # Ask the user for a price
    price = input("Enter price: ")

    # If they type 'done', stop the loop
    if price == "done":
        break

    # Convert what they typed into a number and add it to the total
    total = total + float(price)

# Print the final result
print("The total is: $", total)
```
